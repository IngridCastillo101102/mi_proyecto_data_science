```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. NUMPY: Generar datos numéricos ficticios
np.random.seed(42)  # Para mantener los mismos datos aleatorios siempre
dias = np.arange(1, 11)  # Días del 1 al 10
ventas = np.random.randint(100, 500, size=10)  # 10 valores de ventas aleatorios

# 2. PANDAS: Estructurar los datos en un DataFrame y calcular métricas
df = pd.DataFrame({"Dia": dias, "Ventas_USD": ventas})

# Agregar una columna calculada en Pandas
df["Ventas_Con_IVA"] = df["Ventas_USD"] * 1.16

print("--- Tabla de Datos (Pandas) ---")
print(df.head())

promedio = df["Ventas_USD"].mean()
print(f"\nPromedio de ventas diarias: ${promedio:.2f} USD")

# 3. MATPLOTLIB: Visualizar el comportamiento de las ventas
plt.figure(figsize=(8, 4))
plt.plot(
    df["Dia"],
    df["Ventas_USD"],
    marker="o",
    color="b",
    linestyle="-",
    label="Ventas Diarias",
)
plt.axhline(
    y=promedio,
    color="r",
    linestyle="--",
    label=f"Promedio (${promedio:.1f})",
)

plt.title("Evolución de Ventas - Primeros 10 Días")
plt.xlabel("Día del Mes")
plt.ylabel("Ventas (USD)")
plt.legend()
plt.grid(True)

# Mostrar la gráfica
plt.show()
```
